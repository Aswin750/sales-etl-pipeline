```python
import pandas as pd
print(pd.__version__)
```

```python
df = pd.read_csv("../data/sales_raw_dirty.csv")
duplicate_count = df.duplicated().sum()
print(duplicate_count)
```

```python
print("Shape before cleaning")
print(df.shape)
```

```python
df=df.drop_duplicates()
print("After drop duplicates")
print(df.shape)
print(df.duplicated().sum())
```

```python
df["Customer_Name"]=df["Customer_Name"].fillna("Unknown")
df["Customer_Name"].isnull().sum()
```

```python
price_lookup = {
    "Laptop": 55000,
    "Smartphone": 30000,
    "Headphones": 2500,
    "Keyboard": 1500,
    "Monitor": 12000,
    "Printer": 8000,
    "Mouse": 700,
    "Smartwatch": 5000
}

df["Unit_Price"]=df["Unit_Price"].fillna(df["Product"].map(price_lookup))
df["Unit_Price"].isnull().sum()
```

```python
df["Customer_Name"].head(10)
```

```python
df["Customer_Name"]=df["Customer_Name"].str.strip()
```

```python
df["Payment_Method"].unique()
```

```python
df["Payment_Method"].str.strip().str.title()
df["Payment_Method"]=df["Payment_Method"].replace({"Upi":"UPI"})
df["Payment_Method"].head(10)
```

```python
df["Quantity"].describe()
df[df["Quantity"]<0]
```

```python
df["Discount"].describe()
df[df["Discount"]>100]
```

```python
print("Before cleaning",df.shape)
df=df[df["Quantity"]>0]
df=df[df["Discount"]<=100]
print("After cleaning",df.shape)
print("\nMinimum Quantity")
print(df["Quantity"].min())
print("Maximum discount")
print(df["Discount"].max())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.describe())
```

```python
df.to_csv("../data/cleaned/sales_clean.csv",index=False)
print("Cleaned Successfully")
```
